```python
import pandas as pd
import numpy as np
import json
from glob import glob
import os
os.chdir('..')
```

```python
dataset_type = 'hotel_reviews'
lang = 'indo'
dataset_folders = glob(f'dataset/{dataset_type}/{lang}/*/*.json')
dataset_folders
```

```python
from copy import deepcopy
for json_file in dataset_folders:
	with open(json_file, 'r') as f:
		json_data = json.load(f)
	new_json_data = []
	for item in json_data:
		new_item = deepcopy(item)
		new_item['input'] = new_item['input'].replace('=>', '').strip()
		new_json_data.append(new_item)
	with open(json_file, 'w') as f:
		json.dump(new_json_data, f, indent=4)
	print(f'Processed {json_file}')
```
